```python
import sys; sys.path.insert(0, '../src')
from media_buying_advisor import data, model
P = data.load_params(); ex = P['worked_example']
b = {'S': ex['spend'], 'roas': ex['roas'], 'cpc': ex['cpc'], 'cvr': ex['cvr']}
s = model.scenario(b)
print(f"EPL {s['epl']:.2f}  break-even CPC {s['cvr']*s['epl']:.3f}  profit {s['profit']:,.0f}")
```

```python
chosen = [a for a in P['actions'] if a['id'] in P['plan_ids']]
roas = [model.scenario(b, model.combine(chosen, lv, P['damping']))['roas'] for lv in ('lo','mid','hi')]
print([round(x*100) for x in roas]); assert [round(x*100) for x in roas] == [115,142,173]
```
